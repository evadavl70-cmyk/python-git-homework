```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    posts = response.json()

    for number, post in enumerate(posts[:5], start=1):
        print(f"Пост №{number}")
        print(f"Заголовок: {post['title']}")
        print(f"Текст: {post['body']}")
        print("-" * 40)

except requests.RequestException as error:
    print("Ошибка при получении данных:", error)
```

```python
import requests
from getpass import getpass

api_key = getpass("Введите API-ключ OpenWeather: ")
city = input("Введите название города: ")

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": city,
    "appid": api_key,
    "units": "metric",
    "lang": "ru"
}

try:
    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    weather = response.json()

    temperature = weather["main"]["temp"]
    description = weather["weather"][0]["description"]

    print(f"Город: {weather['name']}")
    print(f"Температура: {temperature} °C")
    print(f"Погода: {description}")

except requests.RequestException as error:
    print("Не удалось получить погоду:", error)
```
